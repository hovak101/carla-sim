```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("training_data/steering.csv")

plt.figure(figsize=(12, 4))
plt.plot(df["frame"], df["steering"], lw=0.8)
plt.xlabel("frame")
plt.ylabel("steering")
plt.title("Training data steering over time")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```
